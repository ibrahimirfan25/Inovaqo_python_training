```python
import threading
import time
import logging
import warnings
import queue
import pdb
from abc import ABC, abstractmethod


try:
    x = 10 / 0
except ZeroDivisionError:
    print("Division by zero handled")


try:
    num = int("abc")
except ValueError:
    print("Invalid integer conversion")


try:
    f = open("sample.txt", "w")
    f.write("Python internship practice")
except:
    print("File error")
finally:
    f.close()


age = -5
if age < 0:
    raise ValueError("Age cannot be negative")


class InvalidMarksError(Exception):
    pass


marks = 120
if marks < 0 or marks > 100:
    raise InvalidMarksError("Marks must be between 0 and 100")


try:
    int("xyz")
except ValueError as e:
    raise RuntimeError("Conversion failed") from e


logging.basicConfig(level=logging.ERROR)
logging.error("Logging an error message")


x = 10
assert x > 0


warnings.warn("This is a warning message")


def simple_task(name):
    time.sleep(1)
    print(name, "completed")


t1 = threading.Thread(target=simple_task, args=("Thread 1",))
t2 = threading.Thread(target=simple_task, args=("Thread 2",))

t1.start()
t2.start()

t1.join()
t2.join()

print("All threads completed")


def background_task():
    while True:
        print("Background running")
        time.sleep(1)


daemon_thread = threading.Thread(target=background_task)
daemon_thread.daemon = True
daemon_thread.start()

time.sleep(3)


counter = 0
lock = threading.Lock()

def increase_counter():
    global counter
    for _ in range(100000):
        with lock:
            counter += 1


t1 = threading.Thread(target=increase_counter)
t2 = threading.Thread(target=increase_counter)

t1.start()
t2.start()
t1.join()
t2.join()

print("Counter value:", counter)


q = queue.Queue()

def producer():
    for i in range(5):
        q.put(i)
        print("Produced:", i)
        time.sleep(1)

def consumer():
    while True:
        item = q.get()
        print("Consumed:", item)
        q.task_done()
        if item == 4:
            break


p = threading.Thread(target=producer)
c = threading.Thread(target=consumer)

p.start()
c.start()

p.join()
c.join()


lock1 = threading.Lock()
lock2 = threading.Lock()

def task_one():
    with lock1:
        time.sleep(1)
        with lock2:
            print("Task one finished")

def task_two():
    with lock1:
        time.sleep(1)
        with lock2:
            print("Task two finished")


t1 = threading.Thread(target=task_one)
t2 = threading.Thread(target=task_two)

t1.start()
t2.start()

t1.join()
t2.join()


stop_event = threading.Event()

def interruptible_task():
    while not stop_event.is_set():
        print("Running task")
        time.sleep(1)
    print("Task stopped safely")


t = threading.Thread(target=interruptible_task)
t.start()

time.sleep(3)
stop_event.set()
t.join()


value = 5
pdb.set_trace()
result = value + 10
print(result)
```
